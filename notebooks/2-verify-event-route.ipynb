```python
import requests
from pprint import pprint

path = "/api/events"
base_url = "http://localhost:8002"
endpoint = f"{base_url}{path}"
endpoint
```

```python
response = requests.get(endpoint)
response.ok
```

```python
if response.ok:
    data = response.json()
    pprint(data)
    print(type(data))
    
```
